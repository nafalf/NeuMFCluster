1. Load Dataset


```python
import pandas as pd

spotify = pd.read_csv(r'D:\TA_NCF+K-means\Code\NCF_Cluster\data\spotify_dataset.csv')

print(spotify.shape)
print(spotify.columns)
spotify.head()
```

2. Feature Selection

```python
cols = [
    'track_name', 'artist_names', 'danceability', 'energy',
    'speechiness', 'acousticness', 'instrumentalness', 'liveness',
    'valence', 'loudness', 'tempo', 'cover_url', 'cover_url_md', 
    'album_name', 'release_date', 'spotify_url', 'preview_url'
]

spotify = spotify[cols].copy()
```

3. Bersihkan Missing Value

```python
# hapus yang kosong di kolom kunci
spotify = spotify.dropna(subset=['track_name','artist_names'])

# isi fitur numerik dengan median
num_cols = spotify.select_dtypes(include='number').columns
for col in num_cols:
    spotify[col] = spotify[col].fillna(spotify[col].median())
```

4. Normalisasi Teks

```python
spotify['track_name'] = spotify['track_name'].astype(str).str.lower().str.strip()
spotify['artist_names'] = spotify['artist_names'].astype(str).str.lower().str.strip()
```

5. Bersihkan tanda baca

```python
import re

def clean_text(x):
    x = re.sub(r'[^a-z0-9\s]', '', x)
    return x

spotify['track_name'] = spotify['track_name'].apply(clean_text)
spotify['artist_names'] = spotify['artist_names'].apply(clean_text)
```

6. Hapus duplikat item

```python
spotify = spotify.drop_duplicates(subset=['track_name','artist_names'])
```

7. Buat item_id

```python
spotify = spotify.reset_index(drop=True)
spotify['item_id'] = spotify.index
```

8. Susun tabel item final

```python
items = spotify[['item_id','track_name','artist_names'] + list(num_cols)].copy()

print(items.shape)
items.head()
```

9. Cek Duplikat

```python
print("Jumlah item unik:", items['item_id'].nunique())
print("Cek duplikat:", items.duplicated(subset=['track_name','artist_names']).sum())
```

10. Normalisasi Fitur Audio dengan Min-Max Scaller

```python
from sklearn.preprocessing import MinMaxScaler

feature_cols = [
    'danceability','energy','speechiness',
    'acousticness','instrumentalness','liveness',
    'valence','loudness','tempo'
]

scaler = MinMaxScaler()
spotify[feature_cols] = scaler.fit_transform(spotify[feature_cols])

spotify[feature_cols].describe()
```

11. Simpan menjadi item_dataset

```python
spotify.to_csv(r'D:/TA_NCF+K-means/Code/NCF_Cluster/data/item_dataset.csv', index=False)
```

### 13. Membuat item cluster yang berjumlah 4.144

```python
import pandas as pd

item_final = pd.read_csv("../data/item_final_4144.csv")
cluster_raw = pd.read_csv("../kmeans_cluster/item_dataset_5f.csv")

item_cluster_4144 = item_final.merge(
    cluster_raw[["track_id", "cluster_5f"]],
    on="track_id",
    how="left"
)

print("Missing cluster:", item_cluster_4144["cluster_5f"].isna().sum())
print("Jumlah item:", item_cluster_4144["item_id"].nunique())
print("Range item:", item_cluster_4144["item_id"].min(), "-", item_cluster_4144["item_id"].max())
print("Cluster unik:", sorted(item_cluster_4144["cluster_5f"].dropna().unique()))

assert item_cluster_4144["cluster_5f"].isna().sum() == 0
assert item_cluster_4144["item_id"].nunique() == 4144
assert item_cluster_4144["item_id"].min() == 0
assert item_cluster_4144["item_id"].max() == 4143

item_cluster_4144["cluster_5f"] = item_cluster_4144["cluster_5f"].astype(int)

item_cluster_4144.to_csv("../data/item_cluster_4144.csv", index=False)

print("Saved: ../data/item_cluster_4144.csv")
```
